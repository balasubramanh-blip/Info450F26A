```python
import pandas as pd
```

```python
df=pd.read_csv("/content/INFO 450 m&m data (Responses) - Form Responses 1 (1).csv")
```

```python
df.head()
```

```python
#df['count'].hist()
df['count'].plot(kind="box")
```

```python
xbar=df['count'].mean()
s=df['count'].std() #standard deviation
print(xbar,s)
```

```python
#control chart
import matplotlib.pyplot as plt
df['count'].plot(marker='o')
ucl=1*s+xbar
lcl=xbar-1*s
plt.axhline(xbar,color='black')
plt.axhline(ucl,color='red')
plt.axhline(lcl,color='red')
```
