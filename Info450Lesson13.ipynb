##Turkey 's 5

```python
min=df['Holding Cost'].min()
max=df['Holding Cost'].max()
Q1=df['Holding Cost'].quantile(0.25) #lower quartile
Q3=df['Holding Cost'].quantile(0.75) #Upper quartile
med=df['Holding Cost'].median() #median
print(min,max,Q1,Q3,med)
#Outliers
Lower=Q1-1.5*(Q3-Q1)
Upper=Q3+1.5*(Q3-Q1)
print(Lower,Upper)
outliers=df[(df['Holding Cost']<Lower) | (df['Holding Cost']>Upper)]
outliers
```
