```python
def celcius_to_fahrenheit(temperatures):
  new_temp=[]
  for temp in temperatures:
    fahrenheit=temp*9/5+32
    new_temp.append(fahrenheit)
  return (new_temp)
```

```python
celsius_temp=[0,1,2,3]
```

```python
max_temp=max(celcius_to_fahrenheit(celsius_temp))
```

```python
product_prices=[10,15,20,25,30]
product_quantities=[100,200,300,400,500]
```

```python
def calculated_revenue(prices,quantities):
  """ This function takes prices and quantities and returns a list of revenues."""
  revenues=[]
  for price,quantity in zip(prices,quantities):
    revenue=price*quantity
    revenues.append(revenue)
  return (revenues)
```

```python
total=sum(calculated_revenue(product_prices,product_quantities))
```
