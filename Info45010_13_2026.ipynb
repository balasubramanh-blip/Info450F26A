```python
import pandas as pd
import scipy.stats as stats
```

```python
df=pd.read_csv('/content/mm_data.csv')
```


The Status Quo - What mars promises
$$H_0:\mu=16$$
Alternative Hypothesis
$$H_a:\mu<16$$

```python
#1 Sample t test for the hyp test
stats.ttest_1samp(df['count'],16,alternative="less") #data, null mean, alternative sign
```
