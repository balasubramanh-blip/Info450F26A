## Problem 1: Count Words With Exactly 6 Letters

Explanation: The software takes a paragraph given by the user and separates it into its words using the split() method. It then continues through the list of words by using the loop and checks the length of each word on its way. Each time a six-lettered word is located, the counter is increased. Finally the program prints the total number of words that were found in the paragraph.

```python
paragraph = "People learn Python quickly when they practice daily and reflect often."

words = paragraph.split()
count = 0

for word in words:
    if len(word) == 6:
        count += 1

print("Words with exactly 6 characters:", count)
```

##Problem 2: Find the Smallest Number (No min)

Explanation: This program takes a series of numbers as input and sets the smallest number to the first number in the list. The program then loops through the numbers. During each loop, the program checks its current number to see if it's the smallest number found so far. This number is updated if a newer, smaller number appears. Finally, once the entire list has been checked, the smallest number is returned.

```python
numbers = [41, 18, 72, 5, 63, 90, 12, 38, 27, 56,
           44, 9, 81, 14, 67, 33, 29, 100, 2, 75,
           61, 48, 20, 88, 10]

smallest = numbers[0]

for number in numbers:
    if number < smallest:
        smallest = number

print("The smallest number is:", smallest)
```

##Problem 3: Count “In-Stock” and “Out-of-Stock” (Ignore Invalid Entries)

Explanation: The system takes the inventory status as its input and goes through each entry in turn. It verifies each entry against the two accepted conditions: In Stock and Out Of Stock. Whenever the program finds a record that matches, it increases the counter accordingly and simply ignores any records that do not conform to either of those two categories. When the end of the loop has been reached, the system displays the total number of goods available and unavailable.

```python
statuses = ["In-Stock", "Out-of-Stock", 0, "In-Stock", "Backorder",
            "Out-of-Stock", "In-Stock", None, "In-Stock", "Damaged"]

in_stock_count = 0
out_of_stock_count = 0

for status in statuses:
    if status == "In-Stock":
        in_stock_count += 1
    elif status == "Out-of-Stock":
        out_of_stock_count += 1

print("In-Stock:", in_stock_count)
print("Out-of-Stock:", out_of_stock_count)
```

##Problem 4: Filter Words That Contain the Letter “a”

Explanation: After the program receives a list of words from the user, it creates a new empty list for storing the results of the process. Next, the program uses a loop to iterate through each word in the input list and check if letter “a” occurs in the given word. If it finds a successful match, it inserts the word into the results list. After all the words have been processed, the program prints out the results list, which includes all the correct words containing letter “a.” It is worth mentioning that one word from the assignment sheet output is missing. “Goal” is a word with letter “a” in it and it should be counted along with “team,” “data,” “plan,” and “value.”

```python
words = ["team", "goal", "win", "data", "model", "plan", "risk", "value"]

words_with_a = []

for word in words:
    if "a" in word:
        words_with_a.append(word)

print("Words containing 'a':", words_with_a)
```

##Problem 5: Average of Valid Numbers (Ignore Missing Values)

Explanation: The program takes an input list which is a combination of numbers and None values. It processes each item in the list using a loop. Whenever a meaningful number is found, it makes a note of that value and increments a count. After going through the complete input list, the value is averaged out to get the answer.

```python
values = [10, None, 30, 40, None, 20]

total = 0
count = 0

for value in values:
    if value is not None:
        total += value
        count += 1

average = total / count

print("Average (ignoring missing values):", average)
```

##Problem 6: Find and Replace a Keyword (Loop-Based)

Explanation: The program reads a paragraph from the user and breaks it down into a list of words. Next, using a loop, it goes through the entire list of words and checks to see if there are any occurrences of the word Excel. If it finds an occurrence, it substitutes the word Excel with the word Python. The process of replacing words continues until the entire list has been traversed through, after which, the list is combined again to form a modified paragraph.

```python
paragraph = "Excel is useful. Many analysts start in Excel before learning Python."

words = paragraph.split()

for i in range(len(words)):
    if words[i] == "Excel":
        words[i] = "Python"

new_paragraph = " ".join(words)

print(new_paragraph)
```
