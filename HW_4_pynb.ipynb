## Problem 1: Count Words With Exactly 6 Letters

Explanation: The program accepts a paragraph from the user and breaks it into separate words using the split() method. It then iterates through that list with a loop, checking the length of each word with a simple conditional statement. Whenever a word measures exactly six characters, the counter increments by one. Once the loop finishes evaluating the entire list, the program outputs the final tally of six-letter words found in the text.

```python
paragraph = "People learn Python quickly when they practice daily and reflect often."

words = paragraph.split()
count = 0

for word in words:
    if len(word) == 6:
        count += 1

print("Words with exactly 6 characters:", count)
```

##Problem 2: Find the Smallest Number (No min)

Explanation: The program accepts a list of numbers as its input and initializes the smallest value using the very first element. From there, it iterates through the rest of the list using a loop. With each pass, an if statement compares the current value against whatever is stored as the minimum. If it encounters a lower value, it updates the variable accordingly. Once the loop finishes checking every item in the list, the program outputs the final result.

```python
numbers = [41, 18, 72, 5, 63, 90, 12, 38, 27, 56,
           44, 9, 81, 14, 67, 33, 29, 100, 2, 75,
           61, 48, 20, 88, 10]

smallest = numbers[0]

for number in numbers:
    if number < smallest:
        smallest = number

print("The smallest number is:", smallest)
```

##Problem 3: Count “In-Stock” and “Out-of-Stock” (Ignore Invalid Entries)

Explanation: The program accepts a list of inventory statuses as its input and iterates through each record using a loop. It checks the value of each entry against two valid states: In-Stock and Out-of-Stock. Whenever a match is found, the program increments the corresponding counter, while quietly skipping any entries that do not match those two labels. Once the loop has evaluated all the records, it prints the final totals for both available and unavailable items.



```python
statuses = ["In-Stock", "Out-of-Stock", 0, "In-Stock", "Backorder",
            "Out-of-Stock", "In-Stock", None, "In-Stock", "Damaged"]

in_stock_count = 0
out_of_stock_count = 0

for status in statuses:
    if status == "In-Stock":
        in_stock_count += 1
    elif status == "Out-of-Stock":
        out_of_stock_count += 1

print("In-Stock:", in_stock_count)
print("Out-of-Stock:", out_of_stock_count)
```

##Problem 4: Filter Words That Contain the Letter “a”

Explanation: The program accepts a list of words as input and initializes a separate empty list to store the filtered results. It then iterates through each entry using a loop, checking whether the letter a appears anywhere in the word. When it finds a match, it appends that word to the new collection. Once the loop has finished processing every word, the program outputs the resulting list containing only the qualifying entries.

```python
words = ["team", "goal", "win", "data", "model", "plan", "risk", "value"]

words_with_a = []

for word in words:
    if "a" in word:
        words_with_a.append(word)

print("Words containing 'a':", words_with_a)
```

##Problem 5: Average of Valid Numbers (Ignore Missing Values)

Explanation: The program accepts a list containing a mix of numbers and None values. It iterates through each element using a loop, checking to filter out any None entries. Whenever it encounters a valid number, it adds that value to a running total and increments a counter. Once it processes the entire list, the program divides that sum by the count of valid entries to find the average, then outputs the final calculation.

```python
values = [10, None, 30, 40, None, 20]

total = 0
count = 0

for value in values:
    if value is not None:
        total += value
        count += 1

average = total / count

print("Average (ignoring missing values):", average)
```

##Problem 6: Find and Replace a Keyword (Loop-Based)

Explanation: The program accepts a paragraph as input and splits it into a list of individual words. It then uses an indexed loop to walk through the list, checking each position to see if the word matches Excel exactly. Whenever it finds a match, it replaces that element with Python. Once the loop has finished checking the entire list, it joins the words back into a single string and outputs the revised text.

```python
paragraph = "Excel is useful. Many analysts start in Excel before learning Python."

words = paragraph.split()

for i in range(len(words)):
    if words[i] == "Excel":
        words[i] = "Python"

new_paragraph = " ".join(words)

print(new_paragraph)
```
